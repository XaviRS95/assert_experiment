```python
import pandas as pd
```

```python
DIR_1 = 'results/'
DIR_2 = 'results_2/'
DIR_3 = 'results_3/'

MODEL_NAMES = ['codegemma_7b', 'deepseek-coder-v2_16b', 'deepseek-coder-v2_236b', 'qwen2.5-coder_32b', 'qwen3-coder_480b']
```

```python
for MODEL_NAME in MODEL_NAMES:

    ORIGINAL_DATASET_1_PATH = f'{DIR_1}{MODEL_NAME}/sv_results_{MODEL_NAME}_1.csv'
    ORIGINAL_DATASET_2_PATH = f'{DIR_1}{MODEL_NAME}/sv_results_{MODEL_NAME}_FULL_AI_2.csv'
    
    SECOND_DATASET_1_PATH = f'{DIR_2}{MODEL_NAME}/sv_results_{MODEL_NAME}_1.csv'
    SECOND_DATASET_2_PATH = f'{DIR_2}{MODEL_NAME}/sv_results_{MODEL_NAME}_2.csv'
    
    FINAL_DATASET_1_PATH = f'{DIR_3}{MODEL_NAME}/sv_results_{MODEL_NAME}_1.csv'
    FINAL_DATASET_2_PATH = f'{DIR_3}{MODEL_NAME}/sv_results_{MODEL_NAME}_2.csv'
    
    
    original_dataset_1 = pd.read_csv(ORIGINAL_DATASET_1_PATH)
    original_dataset_2 = pd.read_csv(ORIGINAL_DATASET_2_PATH)
    
    second_dataset_1 = pd.read_csv(SECOND_DATASET_1_PATH)
    second_dataset_2 = pd.read_csv(SECOND_DATASET_2_PATH)
    
    final_dataset_1 = second_dataset_1.copy()
    final_dataset_1['time(s)'] = original_dataset_1['time(s)']
    final_dataset_2 = second_dataset_2.copy()
    final_dataset_2['time(s)'] = original_dataset_2['time(s)']
    
    !mkdir $DIR_3$MODEL_NAME/

    final_dataset_1.to_csv(f'{DIR_3}{MODEL_NAME}/sv_results_{MODEL_NAME}_1.csv', index=False)
    final_dataset_2.to_csv(f'{DIR_3}{MODEL_NAME}/sv_results_{MODEL_NAME}_2.csv', index=False)
```
