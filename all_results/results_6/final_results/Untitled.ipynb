```python
import pandas as pd
import re
import os

# Regex to extract the module name
module_pattern = r"module\s+(\w+)"

def get_name(text):
    if pd.isna(text):
        return None
    match = re.search(module_pattern, str(text))
    return match.group(1) if match else None

# 1. Load the master dataset
master_df = pd.read_csv('dataset.csv')

# 2. Get all result files (excluding the master)
all_files = [f for f in os.listdir('.') if f.endswith('.csv') and f != 'dataset.csv']

# Create an output directory for the missing modules
output_dir = "missing_modules_results"
os.makedirs(output_dir, exist_ok=True)

for file in all_files:
    print(file)
    current_df = pd.read_csv(file)
    missing_rows = []

    # 3. Row-by-row comparison
    # We zip master['modules'] (the code) and current['original_code']
    for m_code, c_code in zip(master_df['modules'], current_df['original_code']):
        name_master = get_name(m_code)
        name_current = get_name(c_code)

        # Logic: If the names don't match OR current didn't even produce a module name
        if name_master != name_current or name_current is None:
            missing_rows.append(m_code)

    # 4. Save the missing content to a new CSV
    if missing_rows:
        missing_df = pd.DataFrame(missing_rows, columns=['modules'])
        output_filename = f"missing_{file}"
        missing_df.to_csv(os.path.join(output_dir, output_filename), index=False)
        print(f"Saved {len(missing_rows)} missing modules to {output_dir}/{output_filename}")
    else:
        print(f"No missing modules found for {file}.")

print("\nTask complete. Check the 'missing_modules_results' folder.")
```
