```python
import pandas as pd
```

```python
easy_comb = pd.read_csv('case_easy_comb.csv')
easy_seq = pd.read_csv('case_easy_seq.csv')
```

```python
with open('case_fsm.txt', 'r', encoding='utf-8') as file:
    content = file.read()
    sections = []
    separator = '########################################################################################'
    # First, split by the exact separator
    raw_sections = list(set(content.split(separator)))

    # Clean up each section (strip whitespace) and remove empty sections
    for section in raw_sections:
        cleaned_section = section.strip()
        if cleaned_section:  # Only add non-empty sections
            sections.append(cleaned_section)

    fsm = pd.DataFrame(sections, columns=['modules'])
    fsm.to_csv('case_fsm.csv', index=False)
```

```python
case_fsm = pd.read_csv('case_fsm.csv')
case_fsm.shape
```
