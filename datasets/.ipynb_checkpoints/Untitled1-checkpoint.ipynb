```python
import pandas as pd
```

```python
data = pd.read_csv('dataset.csv')
```

```python
data.shape
```

```python
old_module = '''module BORRAME'''
```

```python
new_module = '''module seq_detector_1001_nonoverlap (
    input logic clk,
    input logic rst_n,
    input logic din,
    output logic detected
);

typedef enum logic [2:0] {S0, S1, S10, S100} state_t;
state_t state, next_state;

// State register
always_ff @(posedge clk or negedge rst_n) begin
    case (rst_n)
        1'b0: state <= S0;
        1'b1: state <= next_state;
    endcase
end

// Next state and output logic
always_comb begin
    next_state = state;
    detected = 1'b0;

    case (state)
        S0: begin
            case (din)
                1'b1: next_state = S1;
                1'b0: next_state = S0;
            endcase
        end

        S1: begin
            case (din)
                1'b1: next_state = S1;
                1'b0: next_state = S10;
            endcase
        end

        S10: begin
            case (din)
                1'b1: next_state = S1;
                1'b0: next_state = S100;
            endcase
        end

        S100: begin
            case (din)
                1'b1: begin
                    next_state = S1;   // Start new sequence from last '1'
                    detected = 1'b1;   // Pattern "1001" detected
                end
                1'b0: next_state = S0; // Reset if next bit is 0
            endcase
        end
    endcase
end
endmodule
'''
```

```python
print(data.loc[data['modules'] == old_module, 'modules'])
```

```python
print(data.loc[data['modules'].str.contains('BORRAME', na=False), 'modules'])
```

```python
data.loc[data['modules'] == old_module, 'modules'] = new_module
```

```python
# Find and update rows where modules equals target_module
data.loc[data['modules'].str.contains('module case41', na=False), 'modules'] = new_module
```
